```python

def Fibonacci_Loop(number):
    old = 1
    new = 1
    for itr in range(number - 1):
        tmpval = new
        new = old
        old = old + tmpval
    return new

# tmpval        1   1   2   3
# new       1   1   2   3   4
# old       1   2   3   4   7

def Fibonacci_Loop_Pythonic(number):
    old, new = 1, 1
    for itr in range(number - 1):
        new, old = old, old + new
    return new

# new   1   1   2   3   5 
# old   1   2   3   5   8


Fibonacci_Loop_Pythonic(20)
```

```python
# Given: Three positive integers k, m, and n, representing a population containing k+m+n
# organisms: k individuals are homozygous dominant for a factor, 
# m are heterozygous, 
# and n are homozygous recessive.

# Return: The probability that two randomly selected mating organisms will produce an individual possessing 
# a dominant allele (and thus displaying the dominant phenotype). Assume that any two organisms can mate.

#     het + het
#   
#     |  A | a
#   -----------
#   A | AA | Aa
#   a | Aa | aa


#     het + rec
#   
#     |  a | a
#   -----------
#   A | Aa | Aa
#   a | aa | aa

# all combinations of K+K and K+M would have at least one dominant allele
# 0.75 combinations of M+M would have at least one dominant allele
# 0.5 combinations of M+N would have at least one dominant allele
# 0 combinations of N+N have a dominant allele

from scipy.special import comb

k = 17
m = 25
n = 26

# total number of organisms in the population
tot_pop = k + m + n

# calculate the number of combos that can be made
tot_comb = comb(tot_pop, 2)

# calculate the number of combos that have a dominant allele
val_comb = comb(k,2) + k*m + k*n + 0.5*m*n + 0.75*comb(m,2)

prob = val_comb / tot_comb

print(f'The probability that an individual will have at least one dominant allele:\n{prob}')
```

```python
# Catalan Numbers and RNA Secondary Structures
# Given: An RNA string s having the same number of occurrences of 'A' as 'U' and the same number of occurrences of 'C' as 'G'. The length of the string is at most 300 bp.
# Return: The total number of noncrossing perfect matchings of basepair edges in the bonding graph of s, modulo 1,000,000.
# Hint: Write a function that counts Catalan numbers via dynamic programming. How can we modify this function to apply to our given problem?

# Import the sys module to read input from standard input (or a file redirected to stdin)
# import sys

# Define the modulo value as specified in the problem (1,000,000)
mod = 1000000

def can_pair(a, b):
    """Define a function that checks if two bases can form a valid base pair."""
    return ( a == 'A'and b == 'U') or ( a == 'U' and b == 'A') or ( a == 'C' and b == 'G') or ( a == 'G' and b == 'C')


                                                                  
def catalan_rna(s):
    """Define the main function that computes the number of noncrossing perfect matchings."""
    n = len(s)    # Get the length of the RNA string
    if n % 2 != 0:
        return 0
    # Create a 2D DP table where dp[i][j] stores the number of ways for substring s[i..j]
    # Initialize all entries to 0
    dp = [[0] * n for _ in range(n)]
    # Base case: empty substrings (or single positions considered empty after pairing) have 1 way
    # We will fill dp[i][i-1] conceptually as 1, handled by loops later
    # Fill the DP table for increasing substring lengths (only even lengths matter)
    for length in range(2, n + 1, 2):    # length is the size of the substring (must be even)
        
        for i in range(n - length + 1):    # Iterate over all possible starting positions i for this length
            j = i + length - 1    # Compute the ending position j of the current substring
            ways = 0   # Initialize the number of ways for this substring to 0
            
            # Try pairing the base at i with every possible complementary base at position k
            # k must be such that the distance allows perfect matching (k - i is odd)
            for k in range(i + 1, j + 1, 2):
                
                if can_pair(s[i], s[k]):    # Check if s[i] and s[k] can form a valid base pair
                    left = 1 if k == i + 1 else dp[i + 1][k - 1]    # Number of ways for the substring strictly between i and k
                    right = 1 if k == j else dp[k+ 1][j]    # Number of ways for the substring strictly after k up to j
                    ways = (ways + left * right) % mod    # Multiply the left and right ways and add to total (modulo MOD)
            
            dp[i][j] = ways    # Store the computed number of ways for substring s[i..j]
    
    return dp[0][n - 1]    # Return the number of ways for the entire string s[0..n-1]



# rna = sys.stdin.read().strip()    # Read the entire input from standard input and strip any leading/trailing whitespace
# Jupyter generally can't use sys,stdin because it is waiting for EOF signal, which is difficult to provide in Jupyter
# user_input = input("Enter data: ")    # Jupyter can natively use this command to read user input
rna = 'AUGAGACGCGCUAUAGUGCCGCAUUACAGUACUAAUUAAUUCGGUCUAGUAUAAUAUCAUGGGCGCAUAUGAUCAUCGCAUCGAUUAUACCGCGCGAAGCUAAUUCGAUACGUAUAAUGCUAAUGCACGGCGCCGGGCUAGCGCGCCUGGUACAUACGUCGCGCGAUCGCUUCGACACAUGCGUCGGGCUAAUAUGGCCUGCAUGCUCCGAUGGACCGGUCGUUAGCAUACGUACAACACCGGUGGCUAUUAUCCACGUAUAUGCGAUGCAUGAACGCCGGUUAUA'    # stdin generally does not work directly in jupyter notebook
result = catalan_rna(rna)    # Compute the result using the catalan_rna function
print(result)    # Print the final answer (the number of noncrossing perfect matchings modulo 1,000,000)
```

```python
# Error Correction in Reads

# Given: A collection of up to 1000 reads of equal length (at most 50 bp) in FASTA format. Some of these reads were generated with a single-nucleotide error. For each read s
#  in the dataset, one of the following applies:
# 1) s was correctly sequenced and appears in the dataset at least twice (possibly as a reverse complement);
# 2) s is incorrect, it appears in the dataset exactly once, and its Hamming distance is 1 with respect to exactly one correct read in the dataset (or its reverse complement).
# Return: A list of all corrections in the form "[old read]->[new read]". (Each correction must be a single symbol substitution, and you may return the corrections in any order.)

from collections import Counter

def read_fasta(data):
    """Parse fasta file and return a list of DNA strings (without headers)"""
    dna_strings = []
    current = ''
    for line in data.strip().splitlines():
        if line.startswith('>'):
            if current:
                dna_strings.append(current)
            current = ''
        else:
            current += line.strip()
    if current:
        dna_strings.append(current)
    return dna_strings



def reverse_complement(seq):
    """Define a function to compute the reverse complement of a DNA sequence."""
    complement = str.maketrans('ATCG', 'TAGC') # Create a translation table that maps each base to its complement
    return seq.translate(complement)[::-1] # Translate the sequence using the complement table and reverse the resulting string



def hamming_distance(s1, s2):
    """Define a function to compute the Hamming distance between two strings of equal length."""
    return sum( c1 != c2 for c1, c2 in zip(s1, s2)) # Sum 1 for every position where the characters in the two strings differ, second part is tuple unpacking for the zip



def correct_reads(fasta_data):
    """Define the main function that solves the error-correction problem."""
    reads = [] # Initialize an empty list to store the raw DNA reads extracted from the FASTA input
    current_seq = '' # Initialize an empty string that will accumulate the current sequence while parsing
    for line in fasta_data.strip().splitlines(): # Split the input into individual lines for processing
        if line.startswith('>'): # Check whether the current line is a FASTA header (starts with '>')
            if current_seq: # If we already have a sequence accumulated, store it
                reads.append(current_seq) # Append the completed sequence to the list of reads
                current_seq = '' # Reset the accumulator for the next sequence
        else: # Otherwise the line contains sequence data
            current_seq += line.strip() # Append the current line (stripped of whitespace) to the accumulating sequence
    if current_seq: # After the loop finishes, append the last sequence if one remains
        reads.append(current_seq) # Store the final sequence

    count = Counter() # Create a Counter that will record how many times each sequence (or its RC) appears
    for read in reads: # Iterate over every read that was extracted
        rc = reverse_complement(read) # Compute the reverse complement of the current read
        canonical = min(read, rc) # Choose a canonical representative: the lexicographically smaller of the read and its RC
        count[canonical] += 1 # Increment the count of the chosen canonical form

    correct = set() # Build a set of all sequences that are considered correct (appear at least twice)
    for seq, freq in count.items(): # Iterate over every unique canonical sequence and its frequency
        if freq >= 2: # A sequence is correct when it (or its reverse complement) appears at least twice
            correct.add(seq) # Add both the sequence itself and its reverse complement to the correct set
            correct.add(reverse_complement(seq)) # add is used instead of append because add is for unique sets

    corrections = [] # Initialize a list that will hold the correction strings of the form "old->new"
    for read in reads: # Iterate over every original read from the dataset
        if read in correct: # Skip reads that are already known to be correct
            continue # Continue to the next read without further processing
        for good in correct: # For an incorrect read, search for the unique correct read that differs by exactly one base
            if hamming_distance(read, good) == 1: # Compute the Hamming distance between the incorrect read and the candidate correct read
                corrections.append(f"{read}->{good}")
                break # Once the matching correct read is found we can stop searching

    return corrections # Return the complete list of correction



# Read the input file produce a string
data_dir = r'C:\Users\ianj_\OneDrive\Documents\coding_projects\rosalind\fasta_data.txt'
with open(data_dir, 'r') as file:
    #data = [line.strip() for line in file.readlines() if line.strip()] #create a list with an element for each line
    data = file.read() # Create one string for entire file, lines separated by "\n"
            
result = correct_reads(data) # Call the correction function on the sample data
for corr in result: # Print each correction on its own line
    print(corr) # Output the correction string
```

```python
# Speeding Up Motif Finding
# Given: A DNA string s (of length at most 100 kbp) in FASTA format.
# Return: The failure array of s.

data_dir = r'C:\Users\ianj_\OneDrive\Documents\coding_projects\rosalind\motif_data.txt'
with open(data_dir, 'r') as file:
    data = file.read().strip().splitlines()

# Extract the DNA sequence by skipping the FASTA header line (the first line starting with '>')
seq = ''.join( line.strip() for line in data if not line.startswith('>'))   # Join all non-header lines into one continuous DNA string, stripping whitespace

# Get the length of the DNA string
n = len(seq)

# Initialize the failure array with zeros; index 0 is always 0 by convention
failure = [0] * n

# Initialize the length of the previous longest proper prefix that is also a suffix
length = 0

# Start from the second character (index 1)
i = 1

# Loop until we have processed the entire string
while i < n:
    # If the current character matches the character at the current prefix length
    if seq[i] == seq[length]:
        # Increase the length of the matching prefix-suffix
        length += 1
        # Store the length in the failure array at position i
        failure[i] = length
        # Move to the next character
        i += 1
    # If characters do not match
    else:
        # If there is a previous match length greater than 0
        if length != 0:
            # Fall back to the failure value of the previous match
            length = failure[length - 1]
        # If no previous match exists
        else:
            # Set failure value to 0
            failure[i] = 0
            # Move to the next character
            i += 1

# Convert each integer in the failure array to a string
failure_str = [str(x) for x in failure]

# Join the string representations with spaces and print the result
print(" ".join(failure_str))
```

```python
# Finding a Shared Spliced Motif
# Given: Two DNA strings s and t (each having length at most 1 kbp) in FASTA format.
# Return: A longest common subsequence of s and t. (If more than one solution exists, you may return any one.)



data_dir = r'C:\Users\ianj_\OneDrive\Documents\coding_projects\rosalind\shared_motif.txt'
with open(data_dir, 'r') as file:
    data = file.read().strip().splitlines()



def parse_fasta(data):
    sequences = []
    current = []
    for line in data:
        if line.startswith('>'):    # Check whether the current line is a FASTA header (starts with '>')
            if current:    # If we already have sequence lines stored, a previous sequence is complete
                sequences.append(''.join(current))    # Join those lines into one string and store it
                current = []    # Reset the temporary list for the next sequence
        else:   # The line is not a header, so it contains sequence characters
            current.append(line.strip())    # Add the cleaned sequence characters to the temporary list
    if current:    # After the loop, handle the final sequence that has no following header
        sequences.append(''.join(current))    # Join and store the last sequence
    return sequences    # Return the list containing the two DNA strings



def longest_common_subsequence(s, t):
    n = len(s)    # Store the length of the first DNA string
    m = len(t)    # Store the length of the second DNA string
    dp = [[0]*(m+1) for _ in range(n+1)]    # Create a (n+1) by (m+1) table initialized with zeros for DP lengths
    for i in range(1, n+1):    # Iterate over each character position of s (1-based for the table)
        for j in range(1, m+1):    # Iterate over each character position of t (1-based for the table)
            if s[i-1] == t[j-1]:    # If the current characters of s and t match
                dp[i][j] = dp[i-1][j-1]+1    # Extend the LCS length from the previous diagonal cell by 1
            else:    # The characters do not match
                dp[i][j] = max(dp[i-1][j],dp[i][j-1])    # Take the longer of the LCS lengths from above or left
    lcs = []    # Create an empty list that will collect characters of the LCS in reverse order
    i, j = n, m    # Start reconstruction from the bottom-right corner of the DP table
    while i>0 and j>0:    # Continue until we reach the top or left border of the table
        if s[i-1] == t[j-1]:    # If the characters that produced this cell match
            lcs.append(s[i-1])    # Add that matching character to the LCS
            i-=1    # Move diagonally up-left
            j-=1    # Move diagonally up-left
        elif dp[i-1][j] >= dp[i][j-1]:    # Prefer moving up when the lengths are equal or larger
            i-=1    # Move up (discard character from s)
        else:    # The left cell has a strictly larger value
            j-=1    # Move left (discard character from t)
    return ''.join(reversed(lcs))    # Reverse the collected characters and join them into the final LCS string



sequences = parse_fasta(data)
s, t = sequences[0], sequences[1]
result = longest_common_subsequence(s, t)
print(result)
```

```python
# Maximum Matchings and RNA Secondary Structures
# Given: An RNA string s of length at most 100.
# Return: The total possible number of maximum matchings of basepair edges in the bonding graph of s.

# Import the math module to use factorial and other mathematical functions
import math

# Define a function that computes the number of maximum matchings for an RNA string
def maximum_matchings(rna):
    # Count the number of adenine (A) bases in the RNA string
    nA = rna.count('A')
    # count the number of uracil (U) bases in the RNA string
    nU = rna.count('U')
    # Count the number of cytosine (C) bases in the RNA string
    nC = rna.count('C')
    # Count the number of guanine (G) bases in the RNA string
    nG = rna.count('G')

    # Determine the maximum number of A-U pairs possible (limited by the scarcer base)
    au_pairs = min(nA, nU)
    # Determine the maximum number of C-G pairs possible (limited by the scarcer base)
    cg_pairs = min(nC, nG)

    # Calculate the number of ways to form the maximum A-U matchings:
    # Choose au_pairs from the more abundant base and assign partners from the scarcer one
    # This is equivalent to P(max(nA, nU), au_pairs) = max! / (max - au_pairs)!
    if nA >= nU:
        # More (or equal) A's than U's: permute A's for the U's
        au_ways = math.factorial(nA) // math.factorial(nA - au_pairs)
    else:
        # More U's than A's: permute U's for the A's
        au_ways = math.factorial(nU) // math.factorial(nU - au_pairs)

    # Calculate the number of ways to form the maximum C-G matchings analogously
    if nC >= nG:
        cg_ways = math.factorial(nC) // math.factorial(nC - cg_pairs)
    else:
        cg_ways = math.factorial(nG) // math.factorial(nG - cg_pairs)

    # The total number of maximum matchings is the product of the independent AU and CG contributions
    return au_ways * cg_ways

# Read the RNA string from standard input (typical for Rosalind-style problems)
s = input().strip()

# Compute the result by calling the function with the input RNA string
result = maximum_matchings(s)

# Print the final answer (the total number of maximum matchings)
print(result)





```
